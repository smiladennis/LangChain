```python
#pip install -U "langchain<1.0" langchain-community langchain-google-genai tavily-python

import os
from dotenv import load_dotenv

from langchain_core.prompts import PromptTemplate
from langchain_classic.chains.llm import LLMChain
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import Tool
from langchain_community.tools import TavilySearchResults
from langchain.agents import initialize_agent,AgentType
```

```python
# 🔐 Load API key from .env file
load_dotenv(dotenv_path=".env")
google_api_key = os.getenv("")
```

```python
# ⚡ Initialize the LLM
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",   # You can change to gemini-1.5-pro, etc.
    temperature=0,
    google_api_key=""
)
```

```python
# Prompt + chain
qa_prompt = PromptTemplate.from_template("Answer clearly: {question}")
qa_chain = LLMChain(llm=llm, prompt=qa_prompt)
```

```python
# Wrap as Tool
qa_tool = Tool(
    name="Simple QA",
    func=qa_chain.run,
    description="Answer factual questions clearly"
)
```

```python
# Initialize Agent
agent=initialize_agent(
    tools=[qa_tool],
    llm=llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True
)
```

```python
response=agent.run("What is LangGraph in LangChain?")
print("Agent Answer :\n",response)
```
