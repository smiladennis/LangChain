```python
# 📌 Install required packages first:
# pip install langchain langchain-google-genai python-dotenv

import os
from dotenv import load_dotenv

from langchain_core.prompts import PromptTemplate
from langchain_classic.chains.llm import LLMChain
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import Tool

# 🔐 Load API key from .env file
load_dotenv()
google_api_key = os.getenv("")

# ⚡ Initialize the LLM
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",   # You can change to gemini-1.5-pro, etc.
    temperature=0,
    google_api_key=""
)

# 📝 Define a prompt template
prompt = PromptTemplate.from_template("Answer clearly: {question}")

# 🔗 Create the LLMChain
qa_chain = LLMChain(llm=llm, prompt=prompt)

qa_tool = Tool(
    name="Simple QA",
    func=qa_chain.run,
    description="A basic LLM chain that answers clear questions"
)


# 🚀 Run a query
query = "What is LangGraph in LangChain?"
response = qa_tool.run({"question": query})


# 🖨️ Output
print("\nUser Question:", query)
print("\nLLM Answer:", response)


```
